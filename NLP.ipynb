```python
#1. Tokenization: Split a sentence into words.
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

text = "Natural Language Processing is interesting."
tokens = word_tokenize(text)

print(tokens)
```

```python
#2. Remove Stop Words

import nltk
nltk.download('stopwords')
nltk.download('punkt')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "This is a simple NLP example for students"
tokens = word_tokenize(text)
#print(tokens)

filtered = [w for w in tokens if w.lower() not in stopwords.words('english')]

print(filtered)
```

```python
#3. Stemming

from nltk.stem import PorterStemmer

ps = PorterStemmer()
words = ["playing", "played", "plays"]

for w in words:
    print(w, "->", ps.stem(w))
```

```python
#4. Lemmatization
import nltk
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
words = ["running", "better", "mice"]

for w in words:
    print(w, "->", lemmatizer.lemmatize(w))
```

```python
#5. Word Frequency Count
from collections import Counter
import nltk
#nltk.download('punkt')

text = "NLP is fun and NLP is useful"
words = nltk.word_tokenize(text.lower())

freq = Counter(words)
print(freq)
```

```python
#6. Part-of-Speech (POS) Tagging
import nltk

# Download required resources
#nltk.download('punkt')
#nltk.download('punkt_tab')
#nltk.download('averaged_perceptron_tagger_eng')

text = "RNSIT teaches Data Science"

# Tokenize
tokens = nltk.word_tokenize(text)

# POS tagging
tags = nltk.pos_tag(tokens)

print(tags)
```

```python
#7a. Simple Sentiment Analysis
from textblob import TextBlob

text = "I love learning Natural Language Processing"
blob = TextBlob(text)

print("Polarity:", blob.sentiment.polarity)

if blob.sentiment.polarity > 0:
    print("Positive Sentiment")
elif blob.sentiment.polarity < 0:
    print("Negative Sentiment")
else:
    print("Neutral Sentiment")
```
